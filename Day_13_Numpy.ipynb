# 12 Find duplicate rows in a NumPy matrix.

```python
import numpy as np
arr = np.array([[1, 2, 3],
    [4, 5, 6],
    [1, 2, 3],
    [7, 8, 9],
    [4, 5, 6],
    [10, 11, 12]])
print(arr)
```

```python
rows , counts = np.unique(arr, axis=0, return_counts=True)
duplicate = rows[counts>1]
duplicate
```
