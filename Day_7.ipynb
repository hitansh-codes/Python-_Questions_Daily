# 17. Convert a list of tuples into a dictionary.

```python
tup = [("Name","Hitansh"),("Age",21),("Profession","Data Scientist")]

d = {}
for key , value in tup:
    d[key] = value
print(d)
```

# 18. Flatten a nested list

```python
l = [[1,2,3],
     [4,5,6],
     [7,8,9]]

new = []

for i in l:
    for n in i:
        new.append(n)
print(new)
```

# 19. Print every element from a nested list

```python
lst = [[10, 20], [30, 40], [50, 60]]

for i in lst:
    for l in i:
        print(l)
```

# 20. Find the sum of all numbers

```python
lst = [[10, 20], [30, 40], [50, 60]]

s = []

for i in lst:
    for l in i:
        s.append(l)
print(sum(s))
    
```
