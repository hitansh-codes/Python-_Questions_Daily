# 10. Replace NaN values with the column mean.

```python
import numpy as np
```

```python
arr = np.array([10,20,np.nan,40,np.nan,60,70,80,np.nan,100])
mean = np.nanmean(arr)
arr[np.isnan(arr)] = mean
arr
```
