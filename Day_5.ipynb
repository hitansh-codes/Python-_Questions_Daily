# 13 .Find the largest and smallest values in a list without using max() or min().

```python
l = [10,5,14,20,100,13,18]


largest = l[0]
smallest = l[0]

for num in l:
    if num > largest:
        largest = num
    
    if num < smallest:
        smallest = num
        
print(largest)
print(smallest)
```

# 14. Sort a list without using sort() or sorted().

```python
l = [10, 2, 3, 5, 100, 7, 4]

new = []

while l:
    smallest = l[0]

    for num in l:
        if num < smallest:
            smallest = num

    new.append(smallest)
    l.remove(smallest)

print(new)
```
