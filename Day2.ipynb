# 7. Check whether a string is a palindrome. 

### palindrome MEANS WHERE LETTER IS SAME FROM BACK AND DRONT EX = LEVEL , RACECAR, MADAM

```python
a = "Hitansh" 

if  a == a[::-1]:
    print(f"Yes{a} is a Palindome string")

else:
    print("No its not")
```

# 8. Count vowels and consonants in a string.  

### Consonents = alphabets which have all the english characters except vowels.

```python
a = "Hitansh soni"

vowels = 0
consonents = 0

for char in a.lower():
    if char in "aeiou":
        vowels += 1
    elif char.isalpha():
        consonents += 1

print(vowels)
print(consonents)
        
```

# 6. Find all numbers that appear more than once.

```python
n =  [1,2,2,3,3,5,5,6,7,8,9,0]
l = [] 

for ch in n:
    if ch not in  l:
        l.append(ch)
    else:
        print(ch ,"appearing more than once")
print(l)
```
