# 4. Remove duplicates from a list while preserving order.

```python
l = [1,1,2,2,33,33,4,4,5,5,6,6,7,7,8,8,9,9,10,10]

new_list = []

for i in l:
    if i  not in new_list:
        new_list.append(i)


print(new_list)
```
